```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import jv, kv
```

```python
n1 = 1.52
n2 = 1.51
a = 5.5   #(size,um)
wavelength = 1050   #nm
```

```python
V  = a*1e-6 * 2 * np.pi / (1050*1e-9) * np.sqrt(1.52**2 - 1.51**2)
print(f"V = {V}")
```

##### $HE_{11}$ or $LP_{01}$
and
##### $HE_{12}$ or $LP_{01}$

##### $\frac{J_0 (pa)}{paJ_1(pa)} = \frac{K_0(qa)}{qaK_1(qa)}$

```python
pa = np.arange(1,6,0.0001)
qa = np.sqrt(V**2 - pa**2)
```

```python
LHS = jv(0,pa)/(pa*jv(1,pa))
RHS = kv(0, qa)/(qa*kv(1,qa))
plt.plot(pa, LHS)
plt.plot(pa, RHS)
plt.ylim(-1,1)
```

```python
diffs = np.abs(LHS - RHS)
for i, diff in enumerate(diffs):
    if diffs[i] < 1e-4:
        print(pa[i])
```

##### $TM_{01}$, $TE_{01}$
and
##### $TM_{02}$, $TE_{02}$

##### $\frac{J_1 (pa)}{paJ_0(pa)} = -\frac{K_1(qa)}{qaK_0(qa)}$

```python
LHS = jv(1,pa)/(pa*jv(0,pa))
RHS = -kv(1, qa)/(qa*kv(0,qa))
plt.plot(pa, LHS)
plt.plot(pa, RHS)
plt.ylim(-2,1)
```

```python
# HE11
LP01_pa = 2.04

# TE01, HE21, TM01
LP11_pa = 3.23

# EH11, HE31, HE12
LP21_pa = 4.30

# HE12
LP02_pa = 4.59

# EH21, HE41
LP31_pa = 5.29

# TE02, TM02, HE22
LP12_pa = 5.64
```

```python
def calculate(pa, n1=1.52, a=5.5*1e-6, wavelength=1050*1e-9):
    neff2 = n1**2 - ((pa*wavelength)/(2*np.pi*a))**2
    beta2 = (n1*2*np.pi/wavelength)**2 - (pa/a)**2
    return np.sqrt(neff2), np.sqrt(beta2)
```

```python
LP01_n, LP01_b = calculate(LP01_pa)
print(f"Mode LP01: neff = {LP01_n:.4f}, beta = {LP01_b:.3e} rad/m")

LP11_n, LP11_b = calculate(LP11_pa)
print(f"Mode LP11: neff = {LP11_n:.4f}, beta = {LP11_b:.3e} rad/m")

LP21_n, LP21_b = calculate(LP21_pa)
print(f"Mode LP21: neff = {LP21_n:.4f}, beta = {LP21_b:.3e} rad/m")

LP02_n, LP02_b = calculate(LP02_pa)
print(f"Mode LP02: neff = {LP02_n:.4f}, beta = {LP02_b:.3e} rad/m")

LP31_n, LP31_b = calculate(LP31_pa)
print(f"Mode LP31: neff = {LP31_n:.4f}, beta = {LP31_b:.3e} rad/m")

LP12_n, LP12_b = calculate(LP12_pa)
print(f"Mode LP12: neff = {LP12_n:.4f}, beta = {LP12_b:.3e} rad/m")
```
